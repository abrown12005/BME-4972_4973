```python
import numpy as np
import matplotlib.pyplot as plt

f_i = 5*10**6
angle = 0
c = 1.54*(10**(-3)/10**(-6))

frequencies = np.linspace(0,5000,5000)
velo = (frequencies*c)/(2*f_i*np.cos(angle))

plt.plot(frequencies,velo,linewidth=2)
plt.ylabel('Velocity (m/s)')
plt.xlabel('Doppler Frequency (Hz)')
plt.xscale("log")
plt.title("Blood Velocity on Doppler Frequency")

print(f'Velocity at 3.2kHz = {velo[3200]:.3f}Hz')
```
