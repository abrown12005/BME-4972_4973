```python
import numpy as np
import matplotlib.pyplot as plt

fs = 50*1000
x = np.arange(0,5,1/fs)
w = 10 * x
y = np.cos(w*x)

plt.plot(x,y)
plt.show()
```

```python
# Binarization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_csv("HW2.csv")
bin_img = data > np.mean(data)

plt.imshow(bin_img)
```

```python
# Median filter
import pandas as pd
import matplotlib.pyplot as plt
from scipy.ndimage import median_filter

data = pd.read_csv("HW2.csv")
med_img = median_filter(data,3)

plt.imshow(med_img)
```

```python
# Convolution filter
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

data = pd.read_csv("HW2.csv")
matr = np.array([[0,1,0],[1,-4,1],[0,1,0]])
med_img = signal.convolve(data,matr)

plt.imshow(med_img)
```
