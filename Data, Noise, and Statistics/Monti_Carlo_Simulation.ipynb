```python
#Part a and importing packages
import numpy as np
import matplotlib.pyplot as plt

simulations = 100000
chance_number = 1/6

dice_rolls = np.random.binomial(6,chance_number,simulations)
prob_a = np.mean(dice_rolls >=1)

plt.hist(dice_rolls,bins=6,density=True)
plt.show()
print(prob_a)
```

```python
#Part b
num_rolls = 10

chance_7 = 1/6

dice_7 = np.random.binomial(num_rolls,chance_7,simulations)
prob_b = np.mean(dice_7==2)

plt.hist(dice_7,bins=6,density=True)
plt.show()
print(prob_b)
```

```python
#Part c
num_rolls = 10

chance_2 = 1/36

dice_2 = np.random.binomial(num_rolls,chance_2,simulations)
prob_c = np.mean(dice_2==2)

plt.hist(dice_2,bins=6,density=True)
plt.show()
print(prob_c)
```
